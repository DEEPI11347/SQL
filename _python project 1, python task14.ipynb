```python
df["date"] = pd.to_datetime(df["date"], errors="coerce")
```

```python
numeric_columns = [
    "debit_amount",
    "credit_amount",
    "tax_amount",
    "discount_amount",
    "balance"
]

for col in numeric_columns:
    df[col] = pd.to_numeric(df[col], errors="coerce")
```

```python
text_columns = [
    "transaction_id",
    "account_name",
    "transaction_type"
]

for col in text_columns:
    df[col] = df[col].astype(str).str.strip()

print("Text columns cleaned successfully!")
```

```python
df = df.drop_duplicates().copy()

print("Duplicate records after cleaning:")
print(df.duplicated().sum())
```

```python
df[numeric_columns] = df[numeric_columns].fillna(0)
```

```python
print(df.isnull().sum())
```

```python
df["Month"] = df["date"].dt.to_period("M").astype(str)

df["Year"] = df["date"].dt.year

df["Net_Cash_Flow"] = (
    df["credit_amount"] - df["debit_amount"]
)

df["Total_Charges"] = (
    df["tax_amount"] + df["discount_amount"]
)
```

```python
total_debit = df["debit_amount"].sum()

print("Total Debit:", total_debit)
```

```python
total_credit = df["credit_amount"].sum()

print("Total Credit:", total_credit)
```

```python
total_tax = df["tax_amount"].sum()

print("Total Tax:", total_tax)
```

```python
total_discount = df["discount_amount"].sum()

print("Total Discount:", total_discount)
```

```python
total_balance = df["balance"].sum()

print("Total Balance:", total_balance)
```

```python
average_debit = df["debit_amount"].mean()

print("Average Debit:", average_debit)
```

```python
average_credit = df["credit_amount"].mean()

print("Average Credit:", average_credit)
```

```python
average_balance = df["balance"].mean()

print("Average Balance:", average_balance)
```

```python
net_cash_flow = (
    df["credit_amount"].sum()
    - df["debit_amount"].sum()
)

print("Net Cash Flow:", net_cash_flow)
```

```python
type_summary = df.groupby("transaction_type").agg(
    Number_of_Transactions=("transaction_id", "count"),
    Total_Debit=("debit_amount", "sum"),
    Total_Credit=("credit_amount", "sum"),
    Total_Tax=("tax_amount", "sum"),
    Total_Discount=("discount_amount", "sum"),
    Average_Balance=("balance", "mean")
).sort_values(
    "Total_Credit",
    ascending=False
)

type_summary
```

```python
account_summary = df.groupby("account_name").agg(
    Number_of_Transactions=("transaction_id", "count"),
    Total_Debit=("debit_amount", "sum"),
    Total_Credit=("credit_amount", "sum"),
    Total_Tax=("tax_amount", "sum"),
    Total_Discount=("discount_amount", "sum"),
    Average_Balance=("balance", "mean")
).sort_values(
    "Total_Credit",
    ascending=False
)

account_summary
```

```python
month_summary = df.groupby("Month").agg(
    Number_of_Transactions=("transaction_id", "count"),
    Total_Debit=("debit_amount", "sum"),
    Total_Credit=("credit_amount", "sum"),
    Total_Tax=("tax_amount", "sum"),
    Total_Discount=("discount_amount", "sum"),
    Average_Balance=("balance", "mean")
).sort_index()

month_summary
```
